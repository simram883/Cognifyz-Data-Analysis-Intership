```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/Dataset .csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

```python
# Dataset ki first 5 rows
display(df.head())

# Total rows aur columns
print("Rows and Columns:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Dataset ki information
df.info()
```

```python
# Check missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())

# Total missing values
print("\nTotal missing values:", df.isnull().sum().sum())

# Percentage of missing values
print("\nMissing values percentage:")
print((df.isnull().sum() / len(df) * 100).round(2))
```

```python
# Find the most common cuisines
cuisine_counts = df["Cuisines"].dropna().str.split(", ").explode()

top_cuisines = cuisine_counts.value_counts().head(10)

print("Top 10 most common cuisines:")
print(top_cuisines)
```

```python
plt.figure(figsize=(10, 6))

top_cuisines.sort_values().plot(
    kind="barh",
    color="skyblue"
)

plt.title("Top 10 Most Popular Cuisines")
plt.xlabel("Number of Restaurants")
plt.ylabel("Cuisine Type")
plt.tight_layout()
plt.show()
```

North Indian cuisine is the most common in the dataset with 3960 entries. Chinese comes second with 2735 entries followed by fast food with 1986 entries



```python
top_cities = df["City"].value_counts().head(10)

print("Top 10 Cities with Most Restaurants:")
print(top_cities)
```

```python
plt.figure(figsize=(10, 6))

top_cities.sort_values().plot(
    kind="barh",
    color="lightgreen"
)

plt.title("Top 10 Cities by Restaurant Count")
plt.xlabel("Number of Restaurants")
plt.ylabel("City")

plt.tight_layout()
plt.show()
```

New Delhi has the highest number of restaurants in the dataset followed by gurgaon and noida.This shows the most of the listed restaurant are located in these cities

```python
price_counts = df["Price range"].value_counts().sort_index()

print("Restaurants by Price Range:")
print(price_counts)
```

```python
plt.figure(figsize=(8,5))

price_counts.plot(
    kind="bar",
    color="orange"
)

plt.title("Restaurant Price Range Distribution")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

Price range 1 has the highest no of restaurant as per the price range increases the no. of restaurant becomes lower

```python
price_percentage = (price_counts / price_counts.sum()) * 100

print(price_percentage.round(2))
```

```python
delivery_counts = df["Has Online delivery"].value_counts()

print("Online Delivery Availability:")
print(delivery_counts)
```

```python
delivery_counts.plot(
    kind="bar",
    color=["skyblue", "orange"],
    figsize=(7, 5)
)

plt.title("Online Delivery Availability")
plt.xlabel("Online Delivery")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

Most restaurant in the database do not offer online delivery

```python
delivery_rating = df.groupby("Has Online delivery")[
    "Aggregate rating"
].mean()

print("Average Rating by Online Delivery:")
print(delivery_rating.round(2))
```

Restaurant that offer online delivery have a higher average rating (3,25) than restaurant that do not offer online delivery(2.47)

```python
print(df[["Restaurant Name", "City", "Latitude", "Longitude"]].head())
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["Longitude"],
    df["Latitude"],
    alpha=0.4
)

plt.title("Geographic Distribution of Restaurants")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.tight_layout()
plt.show()
```

The scatter plot show the geographical distribution of restaurant using latitude and longitude. The points represent restaurant location and their distribution helps us understand where restaurants are concentrated

```python
print("Missing values in Reviews:")
print(df["Rating text"].isnull().sum())

print("\nRating text distribution:")
print(df["Rating text"].value_counts())
```

```python
rating_text_counts = df["Rating text"].value_counts()

plt.figure(figsize=(9, 5))
rating_text_counts.plot(kind="bar", color="skyblue")

plt.title("Distribution of Restaurant Rating Text")
plt.xlabel("Rating Text")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

The analysis shows that the average rating text has the highest number of restaurant with more than 3500 entries.Not ratef and good around 2000 entries each, very good has around 1000 while excellent and poor have fewer entries

```python
print("Total votes:", df["Votes"].sum())
print("Average votes:", round(df["Votes"].mean(), 2))
print("Maximum votes:", df["Votes"].max())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Votes"].dropna(), bins=30)

plt.title("Distribution of Restaurant Votes")
plt.xlabel("Number of Votes")
plt.ylabel("Number of Restaurants")
plt.tight_layout()
plt.show()
```

The histogram shows the distribution of votees across restaurants.The no of votes varies among restaurant and the distribution help identify difference in their popularity

```python
delivery_price = pd.crosstab(
    df["Price range"],
    df["Has Online delivery"]
)

print(delivery_price)

delivery_price.plot(kind="bar", figsize=(8, 5))
plt.title("Price Range vs Online Delivery")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

The graph compares online delivary availability across different price range it helps shows how the number of restaurant offering and not offering online delivery varies by price range

```python
print("Table Booking Availability:")
print(df["Has Table booking"].value_counts())

table_booking = df.groupby("Has Table booking")["Aggregate rating"].mean()
print("\nAverage Rating by Table Booking:")
print(table_booking.round(2))
```

```python
table_booking_counts = df["Has Table booking"].value_counts()

plt.figure(figsize=(7, 5))
table_booking_counts.plot(kind="bar", color="lightblue")

plt.title("Table Booking Availability")
plt.xlabel("Has Table Booking")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

The analysis shows that most restaurant do not offer table booking only a samller number of restaurant provide table booking facailities

```python
table_rating = df.groupby("Has Table booking")["Aggregate rating"].mean()

print("Average Rating by Table Booking:")
print(table_rating.round(2))

table_rating.plot(kind="bar", color="orange", figsize=(7, 5))
plt.title("Table Booking vs Average Restaurant Rating")
plt.xlabel("Has Table Booking")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

Restaurant that offer table booking have a higher average rating then restaurant that do not offer this facalities. This suggest and associate between table booking availability and restaurant ratings but it does not prove that table booking causes higher ratings
